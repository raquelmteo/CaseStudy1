The telecom problem

Suppose that you are a telecom company with three services: TV, internet and telephone. Each service have a billing option: billing_1, billing_2 and billing_3. A customer can subscribe to a combination of services, and depending on his choices he will have to be billed in a way. For example: customer A is subscribed to the internet and telephone service. In our database this customer would have 1 in the columns "internet" and "telephone", and he would have 1 in "billing_2" and "billing_3". However, due to technical and/or human error the billing configuration is not 100% accurate resulting in two customers with the same subscription being billed in different ways.

## Key finding: train/test CRM overlap is nearly zero

Only 6 out of 97,100 unique CRM combinations in the test set 
also appear in the training set (99.99% are unseen combinations).

**Implication for Sprint 2 (Modeling):**
This means we cannot rely on exact-match methods, such as 
Nearest Neighbor lookup or Label Powerset classification, since 
these require test customers to closely match combinations seen 
during training.

Instead, we should use a model that learns general rules from 
individual CRM features (e.g., Random Forest or XGBoost, one 
model per Billing column), since this type of model can still 
predict correctly for combinations it has never seen exactly, 
by combining rules learned from smaller pieces of the data.
