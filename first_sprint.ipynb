The telecom problem

Suppose that you are a telecom company with three services: TV, internet and telephone. Each service have a billing option: billing_1, billing_2 and billing_3. A customer can subscribe to a combination of services, and depending on his choices he will have to be billed in a way. For example: customer A is subscribed to the internet and telephone service. In our database this customer would have 1 in the columns "internet" and "telephone", and he would have 1 in "billing_2" and "billing_3". However, due to technical and/or human error the billing configuration is not 100% accurate resulting in two customers with the same subscription being billed in different ways.

```python
import pandas as pd
import numpy as np
```

```python
pd.set_option("display.max_columns", None)
```

Reading the "train" dataset

```python
df = pd.read_csv("train.csv")
df.head()
```

Separating our columns in three types: id, input and output

```python
id_col = "MSISDN"
crm_cols = [c for c in df.columns if not c.startswith("BIL") and c != id_col]
bil_cols = [c for c in df.columns if c.startswith("BIL")]
```

Transforming column type due to memory performance issue

```python
df[crm_cols] = df[crm_cols].astype(np.uint8)
df[bil_cols] = df[bil_cols].astype(np.uint8)
```

Selecting active columns by removing constant ones

```python
active_crm = [c for c in crm_cols if df[c].min() != df[c].max()]
active_bil = [c for c in bil_cols if df[c].min() != df[c].max()]

print(f"Active CRM Features retained: {len(active_crm)} / {len(crm_cols)}")
print(f"Active Billing Targets retained: {len(active_bil)} / {len(bil_cols)}")
```

```python
X_active = df[active_crm]
Y_active = df[active_bil]
```

```python
dropped_crm = [c for c in crm_cols if c not in active_crm]
print("Removed CRM columns:", dropped_crm)
```

Saving the cleaned dataset

```python
df_clean = df[active_crm + active_bil]

df_clean.to_csv('cleaned_telecom_data.csv', index=False)
```

```python
# Check for missing values in each column
missing_values = df_clean.isnull().sum()
print(missing_values[missing_values > 0])
print("Total missing values in cleaned dataset:", df_clean.isnull().sum().sum())
```

```python
# Check for fully duplicated rows (same MSISDN or same values across all columns)
duplicate_rows = df_clean.duplicated().sum()
print("Number of fully duplicated rows in cleaned dataset:", duplicate_rows)
```

```python
df_clean = df_clean.drop_duplicates()

print(f'Original shape: {df.shape}')
print(f'Cleaned shape: {df_clean.shape}')
```

```python
print("\nTop 3 most common CRM packs (in cleaned data):")
print(df_clean[active_crm].sum().sort_values(ascending=False).head(3))

print("\nTop 3 most common Billing packs (in cleaned data):")
print(df_clean[active_bil].sum().sort_values(ascending=False).head(3))
```

```python
#Compare whether there are matching combinations in test.csv with our cleaned data.
test = pd.read_csv("test.csv")
```

```python
# Compare CRM combinations: does test have combos never seen in train?
train_combos = set(map(tuple, df_clean[active_crm].values))
test_combos = set(map(tuple, test[active_crm].values))

overlap = train_combos & test_combos

print("Unique CRM combos in train:", len(train_combos))
print("Unique CRM combos in test:", len(test_combos))
print("Test combos also seen in train:", len(overlap))
print("Test combos NEVER seen in train:", len(test_combos - train_combos))
```

## Key finding: train/test CRM overlap is nearly zero

Only 6 out of 97,100 unique CRM combinations in the test set 
also appear in the training set (99.99% are unseen combinations).

**Implication for Sprint 2 (Modeling):**
This means we cannot rely on exact-match methods, such as 
Nearest Neighbor lookup or Label Powerset classification, since 
these require test customers to closely match combinations seen 
during training.

Instead, we should use a model that learns general rules from 
individual CRM features (e.g., Random Forest or XGBoost, one 
model per Billing column), since this type of model can still 
predict correctly for combinations it has never seen exactly, 
by combining rules learned from smaller pieces of the data.
