```python
import numpy as np
import pandas as pd

from sklearn.feature_selection import mutual_info_classif
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

import warnings
from pandas.errors import PerformanceWarning

warnings.filterwarnings('ignore', category=PerformanceWarning)
pd.set_option('display.max_columns', None)
```

```python
#1. Building the test set

train = pd.read_csv("train.csv")
test = pd.read_csv("test.csv")
solution = pd.read_csv("solution.csv")
```

```python
ID_COL = "MSISDN"

CRM_COLS = [c for c in train.columns if c.startswith("CRM")]
BIL_COLS = [c for c in train.columns if c.startswith("BIL")]
```

```python
SOLUTION_STRING_COL = "Bill_Conf"

test = test.merge(
    solution[[ID_COL, SOLUTION_STRING_COL]],
    on=ID_COL,
    how="left"
)
```

```python
solutions = test['Bill_Conf'].apply(list).apply(pd.Series).astype(int)

test = pd.concat([test, solutions], axis=1)
```

```python
test = test.drop(columns=['Bill_Conf'])
```

```python
columns_to_rename = [c for c in test.columns if c not in CRM_COLS and c != ID_COL]
```

```python
rename_dict = dict(zip(columns_to_rename, BIL_COLS))

test = test.rename(columns=rename_dict)
```

```python
#2. Define X and Y

X = train[CRM_COLS].astype(np.uint8)
Y = train[BIL_COLS].astype(np.uint8)

X_test = test[CRM_COLS].astype(np.uint8)
Y_test = test[BIL_COLS].astype(np.uint8)

# Convert only the arrays used for calculations to int64.
# This prevents integer overflow when counting observations.
X_np = X.to_numpy(dtype=np.int64)
Y_np = Y.to_numpy(dtype=np.int64)
```

```python
# 3. Binary Mutual Information Filter

MI_TOP_K = 25

n_samples = X_np.shape[0]
n_crm = X_np.shape[1]
n_bil = Y_np.shape[1]

selected_features = {}

# Number of 1s and 0s for every CRM variable
x_ones = X_np.sum(axis=0)
x_zeros = n_samples - x_ones

for j in range(n_bil):

    y = Y_np[:, j]

    y_ones = y.sum()
    y_zeros = n_samples - y_ones

    # Count CRM=1 AND Billing=1 for every CRM variable
    xy_11 = X_np.T @ y

    # Remaining cells of each 2x2 contingency table
    xy_10 = x_ones - xy_11
    xy_01 = y_ones - xy_11
    xy_00 = n_samples - xy_11 - xy_10 - xy_01

    # Expected counts under independence
    expected_11 = (x_ones * y_ones) / n_samples
    expected_10 = (x_ones * y_zeros) / n_samples
    expected_01 = (x_zeros * y_ones) / n_samples
    expected_00 = (x_zeros * y_zeros) / n_samples

    # Exact binary mutual information
    mi = np.zeros(n_crm, dtype=np.float64)

    cells = [
        (xy_11, expected_11),
        (xy_10, expected_10),
        (xy_01, expected_01),
        (xy_00, expected_00)
    ]

    for observed, expected in cells:

        valid = observed > 0

        mi[valid] += (
            observed[valid] / n_samples
        ) * np.log(
            observed[valid] / expected[valid]
        )

    # Keep only the top CRM variables
    k = min(MI_TOP_K, n_crm)

    top_idx = np.argpartition(
        mi,
        -k
    )[-k:]

    top_idx = top_idx[
        np.argsort(mi[top_idx])[::-1]
    ]

    selected_features[BIL_COLS[j]] = [
        CRM_COLS[idx]
        for idx in top_idx
    ]

    if (j + 1) % 50 == 0:
        print(f"MI progress: {j + 1}/{n_bil}")

# Filter summary

print(
    f"\nMI filter completed: "
    f"{n_bil} Billing targets × {n_crm} CRM features"
)

print(
    f"CRM features retained per Billing target: {MI_TOP_K}"
)
```

```python
#4. Wrapper

WRAPPER_TOP_K = 10

for i, bil_col in enumerate(BIL_COLS):

    features = selected_features[bil_col]
    y = Y[bil_col].values

    if len(features) > WRAPPER_TOP_K and np.unique(y).size > 1:

        model = LogisticRegression(
            solver="liblinear",
            max_iter=100,
            random_state=42
        )

        model.fit(
            X[features],
            y
        )

        # Rank the MI-selected features by model importance
        importance = np.abs(model.coef_[0])

        top_idx = np.argsort(
            importance
        )[-WRAPPER_TOP_K:]

        selected_features[bil_col] = [
            features[idx]
            for idx in top_idx
        ]

    if (i + 1) % 50 == 0:
        print(f"Wrapper progress: {i + 1}/{n_bil}")
```

```python
#5. One vs All model

models = {}

for i, bil_col in enumerate(BIL_COLS):

    features = selected_features[bil_col]
    y = Y[bil_col].values

    if np.unique(y).size < 2:

        models[bil_col] = None

    else:

        model = LogisticRegression(
            solver="liblinear",
            max_iter=200,
            random_state=42
        )

        model.fit(
            X[features],
            y
        )

        models[bil_col] = model

    if (i + 1) % 50 == 0:
        print(f"Model progress: {i + 1}/{n_bil}")
```

```python
#6. Prediction

Y_pred = np.zeros(
    (len(X_test), len(BIL_COLS)),
    dtype=np.uint8
)

for j, bil_col in enumerate(BIL_COLS):

    model = models[bil_col]
    features = selected_features[bil_col]

    if model is None:

        Y_pred[:, j] = Y[bil_col].iloc[0]

    else:

        Y_pred[:, j] = model.predict(
            X_test[features]
        ).astype(np.uint8)


Y_pred = pd.DataFrame(
    Y_pred,
    columns=BIL_COLS,
    index=test.index
)
```

```python
#7. Exact Match Ratio

exact_matches = (
    Y_pred.values == Y_test.values
).all(axis=1)

EMR = exact_matches.mean()

print("\n==============================")
print(f"Exact Matching Ratio: {EMR:.6f}")
print(
    f"Correct samples: "
    f"{exact_matches.sum()}/{len(exact_matches)}"
)
print("==============================")
```

```python
#8. SAVE PREDICTIONS

prediction = pd.concat([test[[ID_COL]], Y_pred], axis=1)

prediction.to_csv("billing_predictions.csv", index=False)
```
